### 문제
AEIOU 
사전
-> 중복이 가능한 순열

word로 단어 하나가 들어올때
사전에서 몇 번째 단어인지 return

### 왜 이 알고리즘?
중복이 가능한 순열 문제여서 백트레킹이라 생각

가능한 모든 단어 생성
문자열로 변환 후 사전순 정렬
index()로 word 위치 탐색

### 시간 복잡도
1. 전체 생성되는 단어의 수는 
중복정렬 5^1 + 5^2 + 5^3 + 5^4 + 5^5 = 상수
2. 정렬 sort 함수 = NlogN

### 공간 복잡도
1. result에 전체 단어를 저장했으므로, 
전체 중복조합 단어 저장 수에 비례한다
이는 상수이므로
O(N)

### 엣지 케이스
1. word 길이는 1 이상 5 이하
2. 접두사 관계가 있는 문자의 정렬 -> 문자 길이가 달라도 파이썬 sort에서는, 더 짧은 문자가 앞에 옴) 

### 새로 배운 것
1. product 조합을 만들어주는 iterator
product(리스트, repeat = i)
product는 결과를 튜플로 반환한다 (문자 하나씩)
-> 문자열로 변환

2. append() 하면 여러 결과를 하나의 덩어리로, 하나의 객체가 들어감
-> 각각의 단어를 넣으려면 extend

3. ''.join(합치고 싶은 원소)
4. 리스트이름.index(찾고싶은 value)

```python
from itertools import product 

def solution(word):
    #중복이 가능한 순열

    #카운트 해서
    #해당 단어가 몇번째에 있는지
    #완전탐색으로 찾으면 

    aeiou = ['A','E','I','O','U']

    result = []
    diction = []
    for i in range(1,6):
        result.extend(product(aeiou, repeat = i))
        
    for i in result:
        diction.append(''.join(i))
    
    diction.sort()
    return diction.index(word) + 1

    
```
