### 문제

2. 동점인 경우 어피치가 가져감
3. 최종 점수 동점이면 어피치 우승

10 ~ 0 점까지 1점 차이 나는 과녁
10칸의 리스트가 있고
각각 라이언, 어피치가 과녁을 쏘는데 여기서
한 칸당 그 안에 과녁수가 많은 사람이 점수를 가저간다

라이언이 쏠 차례
결과값 : 라이언이 어피치 큰 점수차로 이기기 위해서, 어느 과녁에 몇개의 화살 맞춰야 하는지
    10점부터 내림차순으로 몇발 맞춰야 하는지 정수 배열에 담아 리턴하기

```python
from itertools import combinations_with_replacement 
from collections import Counter
#리스트, 조합 내의 개수를 세 주는 라이브러리

def calculate_score(combi, info):
    #combi는 라이언이 각 화살에 맞힌 점수들
    score1, score2 = 0,0
    for i in range(1,11): #1점부터 10점
        if info[10-i] < combi.count(i):
            #1점에 맞힌 갯수 점점 내려감 
            score1 += i
        elif info[10-i] > 0:
            score2 += i

    return score1, score2

#최대 차이, 조합 저장 , 필요할때마다 조합 생성 iterator 한번에 다 저장하지 않음
def calculate_best_combination(n, info):
    max_diff = 0
    max_comb = {}

    for combi in combinations_with_replacement(range(11), n):
        cnt = Counter(combi)
        #점수  : 맞힌 화살 수 
        score1, score2 = calculate_score(combi, info)
        diff = score1 - score2
        #양수가 나와야 라이언이 이긴것

        if diff > max_diff:
            max_diff = diff
            max_comb = cnt
    return max_diff, max_comb

def solution(n, info):
    max_diff, max_comb = calculate_best_combination(n, info)
    #max_diff는 최종 차이. 라이언이 커야 우승
    #max_comb 는 조합! 

    if max_diff >0:
        answer = [0]* 11
        for score in max_comb:
            answer[10-score] = max_comb[score]
        return answer
    else: 
        return -1
```
